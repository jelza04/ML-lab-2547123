# Naive Bayes Classifier - Play Tennis Dataset

This notebook builds a **Categorical Naive Bayes** model on the Play Tennis dataset, evaluates it with a train/test split, and compares its performance against a Decision Tree and Logistic Regression.

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import CategoricalNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

## Step 1: Create the dataset

```python
data = {
    'Outlook': ['Sunny','Sunny','Overcast','Rain','Rain','Rain','Overcast','Sunny','Sunny','Rain',
                'Sunny','Overcast','Overcast','Rain','Sunny','Sunny','Sunny','Sunny','Overcast','Overcast',
                'Overcast','Overcast','Rain','Rain','Rain','Rain','Rain','Rain','Rain','Sunny',
                'Sunny','Sunny','Sunny','Sunny','Overcast','Overcast','Overcast','Overcast','Rain','Rain',
                'Rain','Sunny','Sunny','Sunny','Overcast','Overcast','Rain','Rain','Sunny','Overcast'],

    'Temperature': ['Hot','Hot','Hot','Mild','Cool','Cool','Cool','Mild','Cool','Mild',
                     'Mild','Mild','Hot','Mild','Cool','Cool','Hot','Hot','Cool','Cool',
                     'Mild','Mild','Hot','Hot','Hot','Hot','Cool','Cool','Mild','Mild',
                     'Cool','Hot','Hot','Mild','Hot','Hot','Cool','Mild','Mild','Cool',
                     'Hot','Cool','Mild','Hot','Cool','Mild','Mild','Cool','Cool','Hot'],

    'Humidity': ['High','High','High','High','Normal','Normal','Normal','High','Normal','Normal',
                 'Normal','High','Normal','High','High','High','Normal','Normal','High','High',
                 'Normal','Normal','High','High','Normal','Normal','High','High','Normal','High',
                 'Normal','High','Normal','Normal','High','Normal','Normal','High','High','Normal',
                 'High','High','Normal','High','High','Normal','Normal','High','Normal','Normal'],

    'Wind': ['Weak','Strong','Weak','Weak','Weak','Strong','Strong','Weak','Weak','Weak',
             'Strong','Strong','Weak','Strong','Weak','Strong','Weak','Strong','Weak','Strong',
             'Weak','Strong','Weak','Strong','Weak','Strong','Weak','Strong','Strong','Strong',
             'Strong','Weak','Weak','Weak','Strong','Strong','Weak','Weak','Weak','Weak',
             'Strong','Weak','Strong','Strong','Weak','Strong','Weak','Strong','Weak','Weak'],

    'Play Tennis': ['No','No','Yes','Yes','Yes','No','Yes','No','Yes','Yes',
                     'Yes','Yes','Yes','No','No','No','Yes','Yes','Yes','Yes',
                     'Yes','Yes','Yes','No','Yes','No','Yes','No','Yes','No',
                     'Yes','No','Yes','Yes','Yes','Yes','Yes','Yes','Yes','Yes',
                     'No','No','Yes','No','Yes','Yes','Yes','No','Yes','Yes']
}

df = pd.DataFrame(data)

print("Dataset")
print(df)
print("\nShape:", df.shape)
```

## Step 2: Separate input features (X) and target variable (y)

```python
X = df[['Outlook', 'Temperature', 'Humidity', 'Wind']]
y = df['Play Tennis']

print("Features (X)")
print(X.head())

print("\nTarget (y)")
print(y.head())
```

## Step 3: Convert categorical values into numerical values

`CategoricalNB` needs integer-encoded categories, so we use `OrdinalEncoder` to convert each categorical column into numbers.

```python
encoder = OrdinalEncoder()
X_encoded = encoder.fit_transform(X)

print("Encoding used for each feature:")
for col, cats in zip(X.columns, encoder.categories_):
    print(f"  {col}: {list(cats)}")

print("\nEncoded Features (first 5 rows)")
print(X_encoded[:5])
```

## Step 4: Divide the dataset into training and testing sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.3, random_state=42
)

print("Training samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])
```

## Step 5: Train a Categorical Naive Bayes model

```python
nb_model = CategoricalNB()
nb_model.fit(X_train, y_train)

print("Naive Bayes model trained.")
print("Classes:", nb_model.classes_)
```

## Step 6: Predict class labels for the test dataset

```python
y_pred = nb_model.predict(X_test)

print("Predicted labels:", list(map(str, y_pred)))
print("Actual labels   :", list(y_test))
```

## Step 7: Accuracy, confusion matrix, and classification report

```python
acc = accuracy_score(y_test, y_pred)
print("Accuracy of Naive Bayes:", acc)

cm = confusion_matrix(y_test, y_pred, labels=nb_model.classes_)
print("\nConfusion Matrix")
print("Labels order:", list(nb_model.classes_))
print(cm)

print("\nClassification Report")
print(classification_report(y_test, y_pred))
```

## Step 8: Predict for a new weather condition

Outlook = Sunny, Temperature = Cool, Humidity = High, Wind = Strong

```python
new_data = pd.DataFrame({
    'Outlook': ['Sunny'],
    'Temperature': ['Cool'],
    'Humidity': ['High'],
    'Wind': ['Strong']
})

new_encoded = encoder.transform(new_data)

new_prediction = nb_model.predict(new_encoded)
new_probability = nb_model.predict_proba(new_encoded)

print("New Sample: Outlook=Sunny, Temperature=Cool, Humidity=High, Wind=Strong")
print("\nPredicted Class:", new_prediction[0])

print("\nClass Probabilities:")
for cls, prob in zip(nb_model.classes_, new_probability[0]):
    print(f"  {cls}: {prob:.4f}")
```

## Step 9: Compare performance with Decision Tree and Logistic Regression

```python
# Decision Tree
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
dt_pred = dt_model.predict(X_test)
dt_acc = accuracy_score(y_test, dt_pred)

# Logistic Regression
lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train, y_train)
lr_pred = lr_model.predict(X_test)
lr_acc = accuracy_score(y_test, lr_pred)

print("Model Comparison")
print(f"Naive Bayes Accuracy     : {acc:.4f}")
print(f"Decision Tree Accuracy   : {dt_acc:.4f}")
print(f"Logistic Regression Acc. : {lr_acc:.4f}")

comparison_df = pd.DataFrame({
    'Model': ['Categorical Naive Bayes', 'Decision Tree', 'Logistic Regression'],
    'Accuracy': [acc, dt_acc, lr_acc]
}).sort_values('Accuracy', ascending=False).reset_index(drop=True)

print("\nComparison Table")
print(comparison_df)
```

## Observations

- The dataset is small (50 rows, 15 in the test split), so accuracy numbers can shift a fair bit depending on the `random_state` used in the split.
- **Naive Bayes** assumes the features are conditionally independent given the class. Even though that assumption doesn't strictly hold for weather attributes, it still does reasonably well here.
- **Decision Tree** can overfit on small categorical datasets since it keeps splitting on the training data.
- **Logistic Regression** works with the ordinal-encoded numeric features here, but ordinal encoding imposes an artificial order (e.g., Sunny < Overcast < Rain) that doesn't really exist for a linear model, which can hurt its performance compared to Naive Bayes and Decision Tree, which don't rely on the numeric ordering.

the model works, but it's better at predicting "yes, they'll play" than "no, they won't"  mainly because it saw far more "yes" examples during training.
